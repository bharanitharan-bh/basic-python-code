```python
arr = [1, 2, 2, 3, 3, 3, 4]

freq = {}

for num in arr:
    freq[num] = freq.get(num, 0) + 1

print(freq)
```

```python
arr = [1, 2, 3, 2, 4, 3, 5]

freq = {}

for num in arr:
    freq[num] = freq.get(num, 0) + 1

for num in freq:
    if freq[num] > 1:
        print(num, end=" ")
```

```python
arr = [1, 2, 3, 2, 4, 3, 5]

freq = {}

for num in arr:
    freq[num] = freq.get(num, 0) + 1

for num in freq:
    if freq[num] == 1:
        print(num, end=" ")
```

```python
arr = [1, 2, 2, 3, 3, 4, 5]

distinct = set(arr)

print(distinct)
```

```python
arr = [1, 2, 2, 3, 3, 4, 5]

distinct = []

for num in arr:
    if num not in distinct:
        distinct.append(num)

print(distinct)
```

```python
def containsDuplicate(nums):
    return len(nums) != len(set(nums))


print(containsDuplicate([1, 2, 3, 1]))  # True
print(containsDuplicate([1, 2, 3, 4]))  # False
```

```python
def firstRepeating(arr):
    freq = {}

    for num in arr:
        freq[num] = freq.get(num, 0) + 1

    for num in arr:
        if freq[num] > 1:
            return num

    return -1


print(firstRepeating([10, 5, 3, 4, 3, 5, 6]))  # 5
```
