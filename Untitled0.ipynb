```python
arr = [1, 2, 2, 3, 3, 4, 5]

distinct = []

for num in arr:
    if num not in distinct:
        distinct.append(num)

print(distinct)
```

```python
def containsDuplicate(nums):
    return len(nums) != len(set(nums))


print(containsDuplicate([1, 2, 3, 1]))  # True
print(containsDuplicate([1, 2, 3, 4]))  # False
```

```python
def firstRepeating(arr):
    freq = {}

    for num in arr:
        freq[num] = freq.get(num, 0) + 1

    for num in arr:
        if freq[num] > 1:
            return num

    return -1


print(firstRepeating([10, 5, 3, 4, 3, 5, 6]))  # 5
```

two pointers

we will v maintain the two variable to track the different position of arr that is left to right

```python
arr = [1, 2, 3, 4, 5]

left = 0
right = len(arr) - 1

while left < right:
    arr[left], arr[right] = arr[right], arr[left]
    left += 1
    right -= 1

print(arr)
```

```python
arr = [2, 2, 3, 3, 1]
k = 1

for i in range(1, len(arr)):
     if arr[i] != arr[i - 1]:
        arr[k] = arr[i]
        k += 1

print(arr)

```

```python
arr = [0, 2, 3, 1, 0, 2, 0]
k = 0

for i in range(len(arr)):
    if arr[i] != 0:
        arr[k] = arr[i]
        k += 1

for i in range(k, len(arr)):
    arr[i] = 0
print(arr)
```

```python
arr = [100, 200, 300, 400]
k = 2

left = 0
sum, max_sum = 0, 0

for right in range(len(arr)):
    sum += arr[right]
    while right - left + 1 > k:
        sum -= arr[left]
        left += 1
        max_sum = max(max_sum, sum)
print(max_sum)

```
